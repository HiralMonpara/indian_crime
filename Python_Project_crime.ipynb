```python
#Which city has the highest and lowest crime count?
count_city=crime['City'].value_counts()
count_city
```

```python
count_city.plot(kind='bar')
plt.title("Crime Count by City")
plt.xlabel("City")
plt.ylabel("Number of Crimes")
plt.xticks(rotation=80)
plt.tight_layout()
plt.show()
```

```python
highest_city=count_city.idxmax()
highest_city
```

```python
lowest_city=count_city.idxmin()
lowest_city
```

```python
#What is the crime trend year by year?
year_counts = crime.groupby('Year').size()
print(year_counts)
```

```python
plt.plot(year_counts.index, year_counts.values, marker='o',color='red')
plt.title("Year-wise Crime Trend")
plt.xlabel("Year")
plt.ylabel("Number of Crimes")
plt.grid(True)
plt.show()
```

```python
# Which crime type is increasing consistently over years?
crime_year = crime.groupby(['Year','Crime Description']).size().unstack(fill_value=0)
trend = crime_year.diff().sum().sort_values(ascending=False)
print("Most Increasing Crime:", trend.idxmax())
```

```python
# What is the closure rate for each city?
closure_city = crime.groupby('City')['Case Closed'] \
                 .apply(lambda x: (x=='Yes').mean()) \
                 .sort_values(ascending=False)

print(closure_city)
```

```python
# Which crime domain has the lowest closure rate?
closure_domain = crime.groupby('Crime Domain')['Case Closed'] \
.apply(lambda x: (x=='Yes').mean())
print("Lowest Closure Domain:", closure_domain.idxmin())
```

```python
# Which age group is most affected?
bins = [0,18,30,45,60,100]
labels = ['0-18','19-30','31-45','46-60','60+']
crime['Age Group'] = pd.cut(crime['Victim Age'], bins=bins, labels=labels)
age_group_counts = crime['Age Group'].value_counts()
print(age_group_counts)
```

```python
age_group_counts.plot(kind='bar',color='purple')
plt.xlabel("Age Group")
plt.ylabel("Number of Crimes")
plt.title("Crime Distribution by Victim Age Group")
plt.show()
```

```python
#  Which gender is more vulnerable in each crime domain?
gender_domain = crime.groupby(['Crime Domain','Victim Gender']).size()
print(gender_domain)
```

```python
gender_domain.unstack().plot(kind='bar', stacked=True)
plt.xlabel("Crime Domain")
plt.ylabel("Number of Crimes")
plt.title("Crime Domain vs Victim Gender")
plt.show()
```

```python
# Does higher police deployment improve closure rate?
deployment_analysis = crime.groupby('Case Closed')['Police Deployed'].mean()
print(deployment_analysis)
```

```python
# Which city has highest violent crime percentage?
violent = crime[crime['Crime Domain']=='Violent Crime']
violent_ratio = violent['City'].value_counts() / crime['City'].value_counts()
print(violent_ratio.sort_values(ascending=False))
```

```python
# What are the top 3 crimes in each city?
top_crimes = crime.groupby(['City','Crime Description']).size() \
               .groupby(level=0) \
               .nlargest(3)

print(top_crimes)
```

```python
# Which weapon is most used in violent crimes?
violent = crime[crime['Crime Domain']=='Violent Crime']
weapon_counts = violent['Weapon Used'].value_counts()

print("Most Used Weapon:", weapon_counts.idxmax())
```
